```python
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt

tmdb = pd.read_csv("https://raw.githubusercontent.com/alura-cursos/data-science-analise-exploratoria/main/Aula_0/tmdb_5000_movies.csv")
tmdb.head()
```

```python
import matplotlib.pyplot as plt
sns.histplot(tmdb["revenue"])
plt.title("Movie revenue")
```

```python
import matplotlib.pyplot as plt
sns.histplot(tmdb["budget"])
plt.title("Movie budget")
```

```python
tmdb["original_language"].unique()
```

```python
languageCount = tmdb["original_language"].value_counts().to_frame().reset_index()

languageCount.columns = ["original_language", "count"]

sns.barplot(data=languageCount, x="original_language", y="count")

sns.countplot(data=tmdb, x="original_language")
```

```python
languageCount.plot(kind="pie", x="original_language", y="count", labels=languageCount["original_language"])
```

```python
 totalLanguage= tmdb["original_language"].value_counts()
 totalLanguage.head()

```

```python
totalEn = totalLanguage.loc["en"]
totalGeneral = totalLanguage.sum()
rest = totalGeneral - totalEn
rest
```

```python
data = {
    "language": ["english","others"],
    "total": [totalEn,rest]
}

data = pd.DataFrame(data)
data
```

```python
sns.barplot(data=data, x="language", y="total")
```

```python
data.plot(kind="pie", x="language", y="total")
```

```python
othersLanguages = tmdb.query("original_language != 'en'")
```

```python

sns.countplot(data=othersLanguages, x="original_language")
```

```python
totalOtherLanguages = tmdb.query("original_language != 'en'")["original_language"].value_counts(normalize=True)
plt.figure(figsize=(10,5))
sns.countplot(data=othersLanguages, order=totalOtherLanguages.index,x="original_language", stat="percent")
plt.title("Movie other languages")
plt.show()
```
